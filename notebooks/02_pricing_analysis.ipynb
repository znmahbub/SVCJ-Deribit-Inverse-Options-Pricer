# 02 — Pricing analysis

Builds the canonical pricing tables and figures from the untouched **30% test holdout** of both fixed calibration specifications:

- `excel files/canonical/calibration_wls.xlsx`
- `excel files/canonical/calibration_pwls.xlsx`

The notebook reports BTC and ETH coin-unit errors separately; pooled BTC+ETH metrics use USD-equivalent or spread-normalized errors. It also exports snapshot diagnostics, maturity/moneyness buckets, within-spread rates, SVCJ parameter diagnostics, and WLS/PWLS comparison tables.


```python
from __future__ import annotations

import sys
from pathlib import Path

NOTEBOOK_ROOT = Path.cwd().resolve()
if (NOTEBOOK_ROOT / "_lib").exists():
    NOTEBOOK_DIR = NOTEBOOK_ROOT
elif (NOTEBOOK_ROOT / "notebooks" / "_lib").exists():
    NOTEBOOK_DIR = NOTEBOOK_ROOT / "notebooks"
else:
    raise FileNotFoundError(f"Could not locate notebooks/_lib from {NOTEBOOK_ROOT}")

if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

from _lib.common import configure_notebook_display, locate_notebook_paths

PATHS = locate_notebook_paths(NOTEBOOK_DIR)
ROOT = PATHS.project_root
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

configure_notebook_display(max_columns=200, width=200)
print("Project root:", ROOT)

from _lib.stage2_workflow import (
    CANONICAL_PRICING_OUTPUT_SUBDIR,
    canonical_calibration_paths,
    run_pricing_analysis,
)
```

```python
WORKBOOKS = canonical_calibration_paths(ROOT)
for key, path in WORKBOOKS.items():
    if not path.exists():
        raise FileNotFoundError(f"Run 01_run_calibration.ipynb first; missing {key}: {path}")
    print(key.upper(), "->", path)

OUTPUT_ROOT = NOTEBOOK_DIR / CANONICAL_PRICING_OUTPUT_SUBDIR
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
print("Pricing outputs ->", OUTPUT_ROOT)
```

```python
pricing_report = run_pricing_analysis(
    WORKBOOKS,
    OUTPUT_ROOT,
    n_buckets=5,
    make_figures=True,
)
```
